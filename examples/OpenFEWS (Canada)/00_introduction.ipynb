## Setting the Scene

This tutorial focuses on a specific and recent event in the Elbow Watershed in Alberta, Canada. Previous floods in this catchment had caused catastrophic damage to the City of Calgary in 2013.

In this recent June, 2026 event, high flows were caused by the combination of snowmelt and rainfall.  This caused high river advisories in many areas of the Bow Catchment, but fortunately minimal flooding and damage.


![plot](images/elbowwatershed.png)


## Rainfall as a driving and uncertain force

The rainfall over the Elbow was forecasted differently in a variety of forecasting Numerical Weather Prediction (NWP). In this case study, we look at the predictions of a deterministic (GDPS) and ensemble (GEPS, GEFS, IFS) NWP models. We can compare these NWP models to station observations and reanalysis (RDPA, HRPA) products.

We want to see which models predicted the rainfall well, and what the skill is over (lead-) time.

![plot](images/elbowrainfall.png)




## Hydrological Forecasting

We also predict the hydrological response in the catchment using a hydrological model ([Raven](https://raven.uwaterloo.ca/)). The model simulation used observed data was pretty accurate! But we want to see how well it was forecasted!
